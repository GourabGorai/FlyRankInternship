## 1. Ranked actions + reason codes

We translate continuous model predictions and page diagnostics into concrete editorial action playbooks:
- `COMPREHENSIVE_REWRITE`: Priority Score >= 0.70 & Age >= 365d (Pillar assets suffering severe staleness and decay).
- `UPDATE_FACTS_AND_TITLE`: Priority Score >= 0.70 & Age < 365d (High-intent assets needing targeted factual refreshes).
- `CONSOLIDATE_OR_PRUNE`: Moderate priority & impressions < 200 (Thin content cannibalizing crawl budget).
- `MONITOR_ONLY`: Mild priority; no active intervention required yet.
- `NO_ACTION`: Stable or growing assets.

```python
import os, sys, pandas as pd, numpy as np
q_path = 'outputs/refresh_queue.csv'
if not os.path.exists(q_path): q_path = '../../outputs/refresh_queue.csv'
df_queue = pd.read_csv(q_path)
# Handle column name variations smoothly
act_col = 'suggested_action' if 'suggested_action' in df_queue.columns else 'recommended_action'
reason_col = 'final_reason_codes' if 'final_reason_codes' in df_queue.columns else 'reason_code'
print('Suggested Action Breakdown:')
print(df_queue[act_col].value_counts())
print('\nReason Code Breakdown in Top 100:')
print(df_queue.head(100)[reason_col].value_counts())
```

## 3. Human review + the no-go list

**Mandatory Human Pre-Check:**
1. Verify the current SERP intent has not shifted to video or forum discussions.
2. Confirm business relevance (e.g. ensure product described is still sold).
3. Check for recent URL migration or 301 redirects.

**The No-Go List (Never Automate):**
- Never auto-rewrite and auto-publish content without manual editor review.
- Never prune or 404 a page without checking external backlink equity.
